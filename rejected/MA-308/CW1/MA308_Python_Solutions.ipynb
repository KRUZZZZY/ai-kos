## Question 1(iii): Coordinate Search

Perform two steps of coordinate search with learning rate α = 0.15, starting at $x_0 = (2, 2)^T$.

The function is: $f(x, y) = 4x^2 - 2xy + y^2 - 3x + 2y + 7$

### Method (Section 3.1.1 of lecture notes — ZERO-ORDER method)

Coordinate search makes **no use of derivatives**. At each sub-step, for coordinate $m$, we evaluate two candidate points:
$$w^{\text{candidate}} = w \pm \alpha e_m$$
where $e_m$ is the $m$-th standard basis vector. We accept whichever gives the lower function value, provided it improves on the current value. One full **step** cycles through all coordinates ($x$, then $y$).

```python
import numpy as np

def f(x, y):
    """Objective function: f(x,y) = 4x^2 - 2xy + y^2 - 3x + 2y + 7"""
    return 4*x**2 - 2*x*y + y**2 - 3*x + 2*y + 7

alpha = 0.15
x, y  = 2.0, 2.0
n_steps = 2

print(f"Starting point: ({x:.3f}, {y:.3f}),  f = {f(x, y):.3f}")
print()

for step in range(1, n_steps + 1):
    print(f"=== Step {step} ===")

    f_current = f(x, y)
    f_plus    = f(x + alpha, y)
    f_minus   = f(x - alpha, y)

    print(f"  Coordinate x  (current x={x:.3f}, y={y:.3f}, f={f_current:.3f}):")
    print(f"    f({x+alpha:.3f}, {y:.3f}) = {f_plus:.3f}   [x + alpha]")
    print(f"    f({x-alpha:.3f}, {y:.3f}) = {f_minus:.3f}   [x - alpha]")

    best_f_x = min(f_plus, f_minus)
    if best_f_x < f_current:
        x = (x + alpha) if (f_plus <= f_minus) else (x - alpha)
        print(f"    -> Accept: x updated to {x:.3f},  f = {f(x,y):.3f}")
    else:
        print(f"    -> No improvement; x unchanged at {x:.3f}")

    f_current = f(x, y)
    f_plus    = f(x, y + alpha)
    f_minus   = f(x, y - alpha)

    print(f"  Coordinate y  (current x={x:.3f}, y={y:.3f}, f={f_current:.3f}):")
    print(f"    f({x:.3f}, {y+alpha:.3f}) = {f_plus:.3f}   [y + alpha]")
    print(f"    f({x:.3f}, {y-alpha:.3f}) = {f_minus:.3f}   [y - alpha]")

    best_f_y = min(f_plus, f_minus)
    if best_f_y < f_current:
        y = (y + alpha) if (f_plus <= f_minus) else (y - alpha)
        print(f"    -> Accept: y updated to {y:.3f},  f = {f(x,y):.3f}")
    else:
        print(f"    -> No improvement; y unchanged at {y:.3f}")

    print(f"  End of step {step}: ({x:.3f}, {y:.3f}),  f = {f(x, y):.3f}")
    print()

print(f"Final point after {n_steps} steps: ({x:.3f}, {y:.3f})")
print(f"f(final) = {f(x, y):.3f}")
```

## Question 2(ii): Linear Regression Statistics

Compute $\bar{x}$, $\bar{y}$, and the sample correlation coefficient $r_{xy}$ using equation (3.4.5) from the lecture notes.

```python
import numpy as np

x_data = np.array([1.73, 1.87, 1.90, 2.66, 3.43, 4.22])
y_data = np.array([4.61, 4.98, 4.39, 6.59, 7.62, 9.65])
n = len(x_data)

x_bar = np.mean(x_data)
y_bar = np.mean(y_data)

S_xx = np.sum((x_data - x_bar)**2)
S_yy = np.sum((y_data - y_bar)**2)
S_xy = np.sum((x_data - x_bar) * (y_data - y_bar))

r_xy = S_xy / np.sqrt(S_xx * S_yy)

print(f"Sample mean of x:  x̄ = {x_bar:.4f}")
print(f"Sample mean of y:  ȳ = {y_bar:.4f}")
print(f"Pearson correlation coefficient: r_xy = {r_xy:.4f}")
```
